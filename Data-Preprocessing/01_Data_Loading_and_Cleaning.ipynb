```python
df=df.drop_duplicates()
df.shape
```

```python
num_cols=df.select_dtypes(include='number').columns
cat_cols=df.select_dtypes(exclude='number').columns
num_cols,cat_cols
```

```python
# Remove columns that are 100% missing
num_df = df[num_cols].dropna(axis=1, how='all')

imp = SimpleImputer(strategy='median')

num_df = pd.DataFrame(
    imp.fit_transform(num_df),
    columns=num_df.columns
)

num_df.head()
```

```python
print(df.shape)

print(len(num_cols))

print(df.select_dtypes(include='number').shape)

print(num_cols.tolist())

print(df.select_dtypes(include='number').columns.tolist())
```

```python
imp=SimpleImputer(strategy='median')
num_df=pd.DataFrame(imp.fit_transform(df[num_cols]),columns=num_cols)
num_df.head()
```

```python
knn=KNNImputer(n_neighbors=5)
knn_df=pd.DataFrame(knn.fit_transform(df[num_cols]),columns=num_cols)
knn_df.head()
```
