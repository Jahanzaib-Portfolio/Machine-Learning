# Notebook 2: Cleaning & Feature Engineering
Stages 6–10

```python
import pandas as pd
import numpy as np

df=pd.read_excel("customer_data_raw.xlsx")
df=df.drop_duplicates()
df.head()
```

## Stage 6 - Fix Invalid Values

```python
# Negative ages
df[df['Age']<0][['CustomerID','Age']]
```

```python
# Replace negative ages with NaN
df.loc[df['Age']<0,'Age']=np.nan
df['Age'].describe()
```

```python
# Invalid dates
df['InvalidDate']=pd.to_datetime(df['InvalidDate'],errors='coerce')
df['InvalidDate'].head()
```

## Stage 7 - Standardize Categorical Values

```python
df['Gender']=df['Gender'].replace({
'M':'Male','male':'Male',
'F':'Female','female':'Female'
})
df['Gender'].value_counts(dropna=False)
```

```python
df['Country']=df['Country'].replace({
'pakistan':'Pakistan',
'PK':'Pakistan'
})
df['Country'].value_counts()
```

```python
df['City']=df['City'].replace({
'karachi':'Karachi',
'KHI':'Karachi'
})
df['City'].value_counts()
```

## Stage 8 - Convert Data Types

```python
df['SalaryText']=df['SalaryText'].replace(r'k$','000',regex=True)
df['SalaryText']=pd.to_numeric(df['SalaryText'],errors='coerce')
df['SalaryText'].head()
```

```python
df['BooleanText']=df['BooleanText'].map({'TRUE':True,'FALSE':False})
df['BooleanText'].head()
```

```python
df['JoinDate']=pd.to_datetime(df['JoinDate'],errors='coerce',dayfirst=True)
df['LastPurchaseDate']=pd.to_datetime(df['LastPurchaseDate'],errors='coerce')
df[['JoinDate','LastPurchaseDate']].dtypes
```

## Stage 9 - Parse Date Columns

```python
df['JoinYear']=df['JoinDate'].dt.year
df['JoinMonth']=df['JoinDate'].dt.month
df['JoinDay']=df['JoinDate'].dt.day
df['JoinWeekday']=df['JoinDate'].dt.day_name()
df[['JoinYear','JoinMonth','JoinDay','JoinWeekday']].head()
```

```python
today=pd.Timestamp.today()
df['CustomerTenureDays']=(today-df['JoinDate']).dt.days
df[['JoinDate','CustomerTenureDays']].head()
```

## Stage 10 - Feature Engineering

```python
df['AgeGroup']=pd.cut(df['Age'],
bins=[0,18,30,45,60,100],
labels=['Child','Young Adult','Adult','Middle Age','Senior'])
df[['Age','AgeGroup']].head()
```

```python
df['SalaryPerExperience']=df['Salary']/df['ExperienceYears'].replace(0,np.nan)
df[['Salary','ExperienceYears','SalaryPerExperience']].head()
```

```python
df['SpendPerOrder']=df['MonthlySpend']/df['NumOrders'].replace(0,np.nan)
df[['MonthlySpend','NumOrders','SpendPerOrder']].head()
```

```python
df['HighValueCustomer']=np.where(df['MonthlySpend']>30000,'Yes','No')
df['HighValueCustomer'].value_counts()
```
