# Notebook 3: Encoding, Outliers & Scaling
Stages 11–16

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import FunctionTransformer, PolynomialFeatures
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler, MinMaxScaler, RobustScaler, PowerTransformer

df=pd.read_excel("customer_data_raw.xlsx")
df=df.drop_duplicates()
```

## Stage 11 - FunctionTransformer

```python
log_transformer=FunctionTransformer(np.log1p, validate=False)
df['MonthlySpend_Log']=log_transformer.transform(df[['MonthlySpend']])
df[['MonthlySpend','MonthlySpend_Log']].head()
```

## Stage 12 - Polynomial Features

```python
poly=PolynomialFeatures(degree=2,include_bias=False)
poly_data=poly.fit_transform(df[['Age','ExperienceYears']].fillna(0))
poly_df=pd.DataFrame(poly_data,columns=poly.get_feature_names_out(['Age','ExperienceYears']))
poly_df.head()
```

## Stage 13 - Encoding

```python
le=LabelEncoder()
df['Purchased_Label']=le.fit_transform(df['Purchased'])
df[['Purchased','Purchased_Label']].head()
```

```python
ohe=OneHotEncoder(sparse_output=False,handle_unknown='ignore')
city=ohe.fit_transform(df[['City']].fillna('Unknown'))
city_df=pd.DataFrame(city,columns=ohe.get_feature_names_out(['City']))
city_df.head()
```

## Stage 14 - Outlier Detection

```python
Q1=df['MonthlySpend'].quantile(0.25)
Q3=df['MonthlySpend'].quantile(0.75)
IQR=Q3-Q1
outliers=df[(df['MonthlySpend']<Q1-1.5*IQR)|(df['MonthlySpend']>Q3+1.5*IQR)]
len(outliers)
```

```python
z=((df['MonthlySpend']-df['MonthlySpend'].mean())/df['MonthlySpend'].std()).abs()
df[z>3][['MonthlySpend']].head()
```

## Stage 15 - Feature Scaling

```python
num=df[['Age','Salary','MonthlySpend']].fillna(0)
std=pd.DataFrame(StandardScaler().fit_transform(num),columns=num.columns)
std.head()
```

```python
mm=pd.DataFrame(MinMaxScaler().fit_transform(num),columns=num.columns)
mm.head()
```

```python
rb=pd.DataFrame(RobustScaler().fit_transform(num),columns=num.columns)
rb.head()
```

## Stage 16 - Power Transformation

```python
positive=df[['MonthlySpend']].copy()
boxcox=PowerTransformer(method='box-cox')
boxcox_df=pd.DataFrame(boxcox.fit_transform(positive),columns=['MonthlySpend_BoxCox'])
boxcox_df.head()
```

```python
yj=PowerTransformer(method='yeo-johnson')
yj_df=pd.DataFrame(yj.fit_transform(df[['Age']].fillna(0)),columns=['Age_YeoJohnson'])
yj_df.head()
```
